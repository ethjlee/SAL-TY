```python
import json
import tarfile
from pathlib import Path
import os
from fnmatch import fnmatch

import pandas as pd
import s2cell
```

```python
directory = Path("/Volumes/Crucial X9")
salty_tars = [filename for filename in os.listdir(directory) if fnmatch(filename, "salty_data_*.tar")]

print(salty_tars)
```

```python
for file in salty_tars:
    with tarfile.open(directory/file, 'r') as archive:
        with archive.extractfile('completed.csv') as csv_file:
            df = pd.read_csv(csv_file)


print(df)
```
