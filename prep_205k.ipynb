```python
import pandas as pd
import numpy as np
```

```python
completed = pd.read_csv("0-100k_data.csv")
all_data = pd.read_csv("1541875_locations_map.csv", header=None).iloc[:, :2]
print(completed)
print(all_data)
```

```python
mask = all_data.index.isin(completed.iloc[:, 0])
print(mask)
new_data = all_data[~mask]

print(new_data)
```

```python
rng = np.random.default_rng(47293)
print(new_data.index.dtype)
selection = rng.choice(new_data.index.values, size=105000, replace=False)

print(len(selection))
print(selection)
```

```python
new_data = new_data.loc[selection].sort_index()
print(new_data.shape)
print(new_data)
```

```python
new_data = new_data.to_csv("100k-205k_data.csv")
```
