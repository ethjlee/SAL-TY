```python
import pandas as pd
import numpy as np
```

```python
first_batch_coords = pd.read_csv("0-100k_data.csv")
second_batch_coords = pd.read_csv("100k-205k_data.csv")
third_batch_coords = pd.read_csv("205k-300k_data.csv")
all_data = pd.read_csv("1541875_locations_map.csv", header=None).iloc[:, :2]
all_data = pd.concat([all_data.index.to_frame(), all_data], axis=1)
all_data.columns = ['ID', 'X', 'Y']
first_batch_coords.columns = ['ID', 'X', 'Y']
second_batch_coords.columns = ['ID', 'X', 'Y']
third_batch_coords.columns = ['ID', 'X', 'Y']
print(first_batch_coords[:5])
print(second_batch_coords[:5])
print(all_data)
print(all_data.iloc[:5, :])
```

```python
completed = pd.concat([first_batch_coords, second_batch_coords, third_batch_coords], axis=0)
completed = completed.sort_values(by='ID')
completed = completed.reset_index(drop=True)
print(completed)
```

```python
mask = all_data.index.isin(completed.iloc[:, 0])
#print(mask)
new_data = all_data[~mask]
print(new_data)

assert(new_data.shape[0] == all_data.shape[0] - completed.shape[0])
```

```python
#old seed = 47293, 86731
rng = np.random.default_rng(383014)
print(new_data.index.dtype)
selection = rng.choice(new_data.index.values, size=200000, replace=False)

print(len(selection))
print(selection)
```

```python
new_data = new_data.loc[selection].sort_index()
print(new_data.shape)
print(new_data)
```
